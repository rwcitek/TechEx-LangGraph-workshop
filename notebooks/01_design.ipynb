## Part A  —  The state schema

Every agentic system has **one shared state object** that every node reads and updates. Getting this right is half the design.

For our **Trivial Agent**, the state should hold:

- nothing; it's trivial

```python
from typing import TypedDict, Annotated, Literal
from pprint import pprint
```

```python
class TriageState(TypedDict):
    # Nothing
    pass

# Sanity check
initial_state: TriageState = {}
print(initial_state)
```

**Hints**:

- `pass` does nothing

## Part B  —  The graph topology

No nodes, except the implied "START" node and the "END" node, with one sequential edge between them.


```python
# The four agent node names.
NODES = {}

# Static edges — always fire in this order.
STATIC_EDGES = []

ENTRY_POINT = "END"

pprint({
    "nodes": NODES,
    "static_edges": STATIC_EDGES,
    "entry_point": ENTRY_POINT,
})
```

**Hints**:

- The `NODES` set is just the agent names — strings, of which there are none.
- `STATIC_EDGES` are pairs that always fire in order, of which there are none.
